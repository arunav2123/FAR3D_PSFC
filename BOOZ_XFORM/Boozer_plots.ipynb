# Generate FAR3D inputs in Boozer co-ordinates from Booz_x_form

```python
import booz_xform as bx
b = bx.Booz_xform()
b.read_wout("vmec_equilibria/W7-X/Standard_fixed_boundary/wout_W7-X_standard_configuration.nc")
```

```python
# Compute booz_x_form quantities 
b.rmnc

# We can set the desired Fourier resolution:
b.mboz = 54
b.nboz = 32

#The transformation to Boozer coordinates will be run on all flux surfaces by default. 
# If desired, we can select only specific surfaces:
print("Default surfaces:", b.compute_surfs)
b.compute_surfs = [0, 1, 2, 23, 47] # Select specific surface number
print("Selected surfaces:", b.compute_surfs)

#Now, run the boozer calculation.
b.run()

#Save all the output data are now available as numpy arrays:
b.bmnc_b
```

```python
# Plot booz_x_form entities 
fig, axes = plt.subplots(2, 2, figsize=(12, 10), dpi=80)
plt.subplots_adjust(hspace=0.4, wspace=0.4)
#1
plt.sca(axes[0, 0])
bx.surfplot(b, js=4)
#2
plt.sca(axes[0, 1])
bx.symplot(b, log=False, sqrts=True)
#3
plt.sca(axes[1, 0])
bx.symplot(b, legend_args={"fontsize": 9, "loc": "upper left", "framealpha": 1})
#4
plt.sca(axes[1, 1])
bx.symplot(b, B0=False, sqrts=True, log=False, legend_args={"fontsize": 11})
plt.show()

```

```python
# Save results to a boozmn_*.nc NetCDF file:
b.write_boozmn("vmec_equilibria/W7-X/Standard_fixed_boundary/boozmn_li383_1.4m.nc")
```
