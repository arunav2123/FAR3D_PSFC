```python
from coilpy import vmec
import matplotlib.pyplot as plt
from mitim_tools.gs_tools import GEQtools
import numpy as np
import omfit_classes.omfit_eqdsk
from scipy.interpolate import interp1d
from numpy.fft import fft
from netCDF4 import Dataset
import xarray as xr
```

```python
wout_path = 'wout_g1051202011.nc'
ds = xr.open_dataset(wout_path)
rmnc = ds['rmnc'].isel(radius=-1).values  # [mn_mode]
zmns = ds['zmns'].isel(radius=-1).values  # [mn_mode]
rmns = np.zeros_like(rmnc)
zmnc = np.zeros_like(zmns)

# Extract mode indices
xm = ds['xm'].values  # [mn_mode]
xn = ds['xn'].values  # [mn_mode]
# Define theta and zeta arrays
ntheta = 360
theta = np.linspace(0, 2 * np.pi, ntheta)
zeta = np.array([0.0])

# Evaluate boundary using MATLAB-style vmec Fourier logic
def cfunct(theta, zeta, coeff, xm, xn):
    theta = np.atleast_1d(theta)
    zeta = np.atleast_1d(zeta)
    mt = np.outer(xm, theta)
    nz = np.outer(xn, zeta)
    cosmt = np.cos(mt)
    sinmt = np.sin(mt)
    cosnz = np.cos(nz)
    sinnz = np.sin(nz)
    result = np.zeros((len(theta), len(zeta)))
    for k in range(1):  # single surface
        rmn = np.tile(coeff[:, np.newaxis], (1, len(theta)))
        result += (rmn * cosmt).T @ cosnz - (rmn * sinmt).T @ sinnz
    return result[:, 0]

def sfunct(theta, zeta, coeff, xm, xn):
    theta = np.atleast_1d(theta)
    zeta = np.atleast_1d(zeta)
    mt = np.outer(xm, theta)
    nz = np.outer(xn, zeta)
    cosmt = np.cos(mt)
    sinmt = np.sin(mt)
    cosnz = np.cos(nz)
    sinnz = np.sin(nz)
    result = np.zeros((len(theta), len(zeta)))
    for k in range(1):
        zmn = np.tile(coeff[:, np.newaxis], (1, len(theta)))
        result += (zmn * sinmt).T @ cosnz + (zmn * cosmt).T @ sinnz
    return result[:, 0]

# Calculate R and Z boundary
r = cfunct(theta, zeta, rmnc, xm, xn) + sfunct(theta, zeta, rmns, xm, xn)
z = cfunct(theta, zeta, zmnc, xm, xn) + sfunct(theta, zeta, zmns, xm, xn)


# Initialize r and z arrays
rs = np.zeros(ntheta)
zs = np.zeros(ntheta)

# Evaluate sums directly (1D version of cfunct + sfunct logic)
for i in range(len(rmnc)):
    rs += rmnc[i] * np.cos(xm[i] * theta - xn[i] * zeta)
    zs += zmns[i] * np.sin(xm[i] * theta - xn[i] * zeta)
```

```python
# Interpolate EFIT boundary onto uniform theta grid
efit_r = efit_xbdry
efit_z = efit_zbdry

# Compute centroid (approximate magnetic axis) for angle definition
r0 = np.mean(efit_r)
z0 = np.mean(efit_z)
theta_efit = np.arctan2(efit_z - z0, efit_r - r0)

# Sort theta and associated R, Z
sorted_indices = np.argsort(theta_efit)
theta_sorted = theta_efit[sorted_indices]
r_sorted = efit_r[sorted_indices]
z_sorted = efit_z[sorted_indices]

# Resample onto uniform theta grid
ntheta = 360
theta_uniform = np.linspace(0, 2 * np.pi, ntheta, endpoint=False)
r_interp = np.interp(theta_uniform, theta_sorted, r_sorted)
z_interp = np.interp(theta_uniform, theta_sorted, z_sorted)

# Compute Fourier coefficients
mpol = 12  # limit for cleaner comparison
rmnc_fit = np.zeros(mpol + 1)
zmns_fit = np.zeros(mpol + 1)
for m in range(mpol + 1):
    rmnc_fit[m] = (2 / ntheta) * np.sum(r_interp * np.cos(m * theta_uniform))
    if m > 0:
        zmns_fit[m] = (2 / ntheta) * np.sum(z_interp * np.sin(m * theta_uniform))

# Compare to VMEC coefficients (using same m range)
# Extract VMEC rmnc and zmns from mode-matched entries
rmnc_vmec = np.zeros(mpol + 1)
zmns_vmec = np.zeros(mpol + 1)
for i in range(len(xm)):
    m, n = int(xm[i]), int(xn[i])
    if n == 0 and m <= mpol:
        rmnc_vmec[m] = rmnc[i]
        zmns_vmec[m] = zmns[i]

# Plot comparison
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.stem(range(mpol + 1), rmnc_fit, linefmt='r-', markerfmt='ro', basefmt='k-', label='EFIT $R_m$')
plt.stem(range(mpol + 1), rmnc_vmec, linefmt='g-', markerfmt='go', basefmt='k-', label='VMEC $R_m$')
plt.title('Cosine Coefficients of R (rmnc)')
plt.xlabel('m')
plt.ylabel('Amplitude')
plt.legend()

plt.subplot(1, 2, 2)
plt.stem(range(1, mpol + 1), zmns_fit[1:], linefmt='r-', markerfmt='ro', basefmt='k-', label='EFIT $Z_m$')
plt.stem(range(1, mpol + 1), zmns_vmec[1:], linefmt='g-', markerfmt='go', basefmt='k-', label='VMEC $Z_m$')
plt.title('Sine Coefficients of Z (zmns)')
plt.xlabel('m')
plt.ylabel('Amplitude')
plt.legend()

plt.tight_layout()
plt.show()
```
